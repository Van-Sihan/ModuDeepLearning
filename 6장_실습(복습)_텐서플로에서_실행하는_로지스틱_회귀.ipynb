```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

x = np.array([2,4,6,8,10,12,14])
y = np.array([0,0,0,1,1,1,1])

model = Sequential()
model.add(Dense(1, input_dim=1, activation='sigmoid'))

#교차 엔트로피 오차 함수를 이용하기 위해 'binary_crossentropy'로 설정
model.compile(optimizer='sgd', loss='binary_crossentropy')
model.fit(x,y,epochs=1000)

#그래프로 확인
plt.scatter(x,y)
plt.plot(x, model.predict(x), 'r') #'r' = red
plt.show()

#임의의 학습 시간을 집어넣어 합격 예상 확률을 예측
hour=7

#tensorflow는 기본 2차원 구조라서 [hour]만 사용시 오류 발생
input_data = tf.constant([[hour]], dtype=tf.float32)

# 코드 의미: 앞에서 1000번 동안 열심히 학습을 마친 딥러닝 모델(model)에게 새로운 입력값(input_data, 여기서는 7시간)을 집어넣고 결과를 예측해달라고 명령하는 함수
# [0][0]의 의미 : 우리가 원하는 것은 대괄호로 감싸진 2차원 리스트가 아니라, 그 안에 들어 있는 순수한 숫자(확률 값) 하나뿐. 그래서 껍질을 두 번 벗겨내는 과정.
prediction = model.predict(input_data)[0][0]

print("%.f 시간을 공부할 경우, 합격 예상 확률은 %.01f%%입니다."%(hour, prediction *100))
```
