```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
import urllib.request
import os

# 데이터 저장 디렉토리 생성
os.makedirs('./data', exist_ok=True)

# 파일명 철자 오류(Thoraric vs Thoracic)와 여러 브랜치를 모두 대응하는 최종 후보군
urls = [
    "https://raw.githubusercontent.com/gilbutITbook/080324/main/repository/data/ThoraricSurgery3.csv",
    "https://raw.githubusercontent.com/gilbutITbook/080324/main/repository/data/ThoracicSurgery3.csv",
    "https://raw.githubusercontent.com/thebookshong/deeplearning-keras/main/repository/data/ThoraricSurgery3.csv",
    "https://raw.githubusercontent.com/thebookshong/deeplearning-keras/main/repository/data/ThoracicSurgery3.csv",
    "https://raw.githubusercontent.com/thebookshong/deeplearning-keras/main/data/ThoraricSurgery3.csv",
    "https://raw.githubusercontent.com/thebookshong/deeplearning-keras/main/data/ThoracicSurgery3.csv"
]

success = False
for url in urls:
    try:
        print(f"Downloading from: {url}")
        urllib.request.urlretrieve(url, "./data/ThoraricSurgery3.csv")
        print("Successfully downloaded!")
        success = True
        break
    except Exception as e:
        print(f"Failed: {e}")

if not success:
    # 최후의 수단으로 다른 신뢰할 수 있는 데이터 미러에서 가져옵니다.
    try:
        backup_url = "https://raw.githubusercontent.com/taehojo/deeplearning/master/data/ThoraricSurgery3.csv"
        print(f"Trying backup source: {backup_url}")
        urllib.request.urlretrieve(backup_url, "./data/ThoraricSurgery3.csv")
        print("Successfully downloaded from backup!")
        success = True
    except Exception as e:
        raise FileNotFoundError("모든 경로에서 데이터를 다운로드하지 못했습니다. 인터넷 연결 또는 파일명을 확인해 주세요.")

# 데이터셋 불러오기 (delimiter=",": 쉼표를 기준으로 데이터 분리)
Data_set = np.loadtxt("./data/ThoraricSurgery3.csv", delimiter=",")

# [:, 0:16] -> 모든 행의 0~15번째 열(총 16개 환자 진찰 기록)을 입력값 X로 추출
X = Data_set[:, 0:16]

# [:, 16] -> 모든 행의 16번째 인덱스 열(17번째 열인 수술 1년 후 생사 여부 정답)을 출력값 y로 추출
y = Data_set[:, 16]

model = Sequential()

# Dense(30...): 중간 학습층 노드 30개 설정 / input_dim=16: 입력되는 데이터 정보가 16개 / activation='relu': 빠른 학습을 위한 표준 활성화 함수
model.add(Dense(30, input_dim=16, activation='relu'))

# Dense(1...): 최종 출력 노드 1개 / activation='sigmoid': 이진 분류(사망/생존)를 위해 0~1 사이의 확률값으로 변환하여 출력
model.add(Dense(1, activation='sigmoid'))

# loss: 이진 분류 채점 기준(binary_crossentropy) / optimizer: 스마트하게 오차를 줄이는 나침반(adam) / metrics: 사람이 볼 정확도 기준
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# epochs=5: 전체 데이터셋을 5번 반복 학습 / batch_size=16: 데이터를 16개씩 나누어 학습하고 오차 수정 / history: 학습 과정을 기록 보관
history = model.fit(X, y, epochs=5, batch_size=16)
```
